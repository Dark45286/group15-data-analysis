```python
#import of pandas, numpy , matplotlib library
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
##This step allows us to preview the first rows of the dataset and understand its structure
df = pd.read_csv("Crop_recommendation.csv")  
df.head()


```

```python
df.shape
#df.shape returns the dimensions of the DataFrame.
```

```python
df.info()
#It gives a summary of the DataFrame
```

```python
df['soil_fertility_index'] = (
    df['N'] +
    df['P'] +
    df['K']
) / 3
# Create a soil fertility index by averaging the main soil nutrients (N, P, and K)
```


plt.figure()
plt.hist(df['soil_fertility_index'], bins=20)
plt.xlabel("Soil Fertility Index")
plt.ylabel("Frequency")
plt.title("Distribution of Soil Fertility Index")
plt.show()


```python
# Visualize the average soil fertility index for each crop type
plt.figure()
df.groupby('label')['soil_fertility_index'].mean().sort_values().plot(kind='bar')
plt.xlabel("Crop Type")
plt.ylabel("Average Soil Fertility Index")
plt.title("Average Soil Fertility Index by Crop Type")
plt.show()
```

```python
# Display the distribution of the soil fertility index values
plt.figure()
plt.hist(df['soil_fertility_index'], bins=20)
plt.xlabel("Soil Fertility Index")
plt.ylabel("Frequency")
plt.title("Distribution of Soil Fertility Index")
plt.show()
```
